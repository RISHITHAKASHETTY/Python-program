## An example: visualizing data in the notebook ✨

Below is an example of a code cell. We'll visualize some simple data using two popular packages in Python. We'll use [NumPy](https://numpy.org/) to create some random data, and [Matplotlib](https://matplotlib.org) to visualize it.

Note how the code and the results of running the code are bundled together.

```python
from matplotlib import pyplot as plt
import numpy as np

# Generate 100 random data points along 3 dimensions
x, y, scale = np.random.randn(3, 100)
fig, ax = plt.subplots()

# Map each onto a scatterplot we'll create with Matplotlib
ax.scatter(x=x, y=y, c=scale, s=np.abs(scale)*500)
ax.set(title="Some random data, created with JupyterLab!")
plt.show()
```

```python
import random
developers = list(range(2,51))
testers = list(range(51,81))
managers = list(range(81,100))
sample_dev = random.sample(developers, 5)
sample_testers = random.sample(testers, 3)
sample_managers = random.sample(managers, 2)

final_sample = sample_dev + sample_testers + sample_managers
print("Stratified sample:", final_sample)
```

```python
cluster = {
    "Hyderabad": [1,2,3,4,5],
    "pune": [6,7,8,9,10],
    "Chennai":[11,12,13,14,15]
}
select_clusters = ["Hyderabad","Chennai"]
sample = []

for city in select_clusters:
    sample.extend(cluster[city])
    print("Cluster sample:",sample)
    
```

```python
import matplotlib.pyplot as plt

products = ["Pen", "Book", "Bag"]
cp = [10, 50, 200]
sp = [15, 45, 250]

profit = [sp[i] - cp[i] for i in range(len(products))]

plt.bar(products, profit, alpha=0.7)
plt.axhline(y=0, color='red', linestyle='--') 
plt.title("Profit/Loss per Product")
plt.ylabel("₹")
plt.show()

for p, pr in zip(products, profit):
    print(f"{p}: Profit = ₹{pr}" if pr>0 else f"{p}: Loss = ₹{abs(pr)}")
```
